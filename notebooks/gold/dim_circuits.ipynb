```python
from pyspark.sql import functions as F
from config.f1_config import CATALOG, SILVER_SCHEMA, GOLD_SCHEMA
GOLD_TABLE = f"{CATALOG}.{GOLD_SCHEMA}.circuits"
```

```python
df_silver_circuits = spark.read.table(f"{CATALOG}.{SILVER_SCHEMA}.circuits")
```

```python
def build_dim_circuits(df_silver_circuits):
    return(
        df_silver_circuits
        .select(
            "Circuit_Id","Circuit_Name" ,"City","Latitude","Longitude", "Country"
            )
)
df_gold_dim_circuits = build_dim_circuits(df_silver_circuits)


display(df_gold_dim_circuits)
```

```python
df_gold_dim_circuits.write.format("delta")\
    .mode("overwrite")\
    .option("overwriteSchema", "true")\
    .saveAsTable(GOLD_TABLE)
```
